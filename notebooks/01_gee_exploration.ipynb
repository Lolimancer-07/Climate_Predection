# 01. Google Earth Engine Exploration: Bay of Bengal Terrain & Land Cover

This notebook demonstrates how the platform queries Google Earth Engine (GEE) to retrieve:
1. **Digital Elevation Models (SRTM 30m / MERIT 90m)** for coastal bathymetry and slope
2. **Dynamic World (10m)** for land use land cover (built-up, water, vegetation)
3. **Topographic Wetness Index (TWI)** calculation for surface runoff modeling

```python
import ee
import os
import json

# Authenticate and initialize GEE
try:
    ee.Initialize()
    print("✓ Google Earth Engine initialized successfully")
except Exception as e:
    print(f"GEE authentication required: {e}")
```

## 1. Load Puri District AOI & SRTM Elevation

```python
# Define bounding box for Puri District, Odisha, India
puri_aoi = ee.Geometry.BBox(85.7, 19.7, 85.95, 19.9)

# Ingest SRTM 30m DEM
dem = ee.Image('USGS/SRTMGL1_003').clip(puri_aoi)
elevation = dem.select('elevation')

# Calculate terrain slope in degrees
slope = ee.Terrain.slope(elevation)

print("SRTM DEM & Slope layers ready for spatial analysis")
```

## 2. Ingest Dynamic World Real-Time Land Cover

```python
dw = ee.ImageCollection('GOOGLE/DYNAMICWORLD/V1') \
    .filterBounds(puri_aoi) \
    .filterDate('2023-01-01', '2023-12-31') \
    .select('label') \
    .mode() \
    .clip(puri_aoi)

print("Dynamic World land cover composite generated")
```
