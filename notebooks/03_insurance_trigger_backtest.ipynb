# 03. Parametric Insurance Trigger Backtest on Historical Cyclone Records

This notebook backtests parametric insurance policy triggers across Bay of Bengal cyclones:
1. **Trigger Types**: Peak Surge Height, Cumulative Rainfall (48h), Sustained 1-min Wind Speed
2. **Backtest Targets**: Fani (2019), Amphan (2020), Yaas (2021), Gulab (2021)
3. **Audit Hash Verification**: Verifying that HMAC-SHA256 signatures are reproducible

```python
from modeling.insurance_trigger.trigger_engine import evaluate_all_policies, DEMO_POLICIES

# Scenario tests: (Event, District, Surge (m), Rainfall (mm), Wind (km/h))
events = [
    ("FANI-2019", {"surge_height": 1.8, "rainfall_total": 280.0, "wind_speed": 215.0}),
    ("AMPHAN-2020", {"surge_height": 4.5, "rainfall_total": 310.0, "wind_speed": 185.0}),
    ("GULAB-2021", {"surge_height": 0.6, "rainfall_total": 120.0, "wind_speed": 95.0}), # Weaker event
]

for event_id, hazards in events:
    results = evaluate_all_policies(event_id, hazards)
    print(f"\n=== {event_id} ===")
    for r in results:
        status = "🔴 FIRED" if r.triggered else "⚪ NOT MET"
        print(f"{status} | {r.trigger_type:15} | Obs: {r.observed_value:6.1f} vs Thr: {r.threshold_value:6.1f} | Hash: {r.audit_hash[:12]}...")
```
